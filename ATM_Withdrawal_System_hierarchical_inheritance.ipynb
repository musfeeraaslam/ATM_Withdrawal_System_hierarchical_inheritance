```python
class Account:
    def __init__(self, account_number, customer_name, balance):
        self.account_number = account_number
        self.customer_name = customer_name
        self.balance = balance

    def show_details(self):
        print(f"Account Number: {self.account_number}")
        print(f"Customer Name: {self.customer_name}")
        print(f"Balance: Rs. {self.balance:.0f}")

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit Failed: Amount must be greater than 0")
            return
        self.balance += amount
        print(f"Deposit Successful. New Balance: Rs. {self.balance:.0f}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal Failed: Amount must be greater than 0")
            return
        if amount > self.balance:
            print("Withdrawal Failed: Insufficient balance")
            return
        self.balance -= amount
        print(f"Withdrawal Successful. New Balance: Rs. {self.balance:.0f}")

    def calculate_profit(self):
        # Base rule — no profit for a generic account
        return 0


class SavingsAccount(Account):
    def calculate_profit(self):
        # Savings: 5% annual profit on balance
        profit = self.balance * 0.05
        print(f"Savings Profit (5%): Rs. {profit:.0f}")
        return profit


class CurrentAccount(Account):
    def calculate_profit(self):
        # Current accounts earn no profit — they're for daily transactions
        print("Current Account: No profit (transaction-only account)")
        return 0


class BusinessAccount(Account):
    def calculate_profit(self):
        # Business: 7% annual profit on balance
        profit = self.balance * 0.07
        print(f"Business Profit (7%): Rs. {profit:.0f}")
        return profit


# --- Example usage ---
savings = SavingsAccount("SA-1001", "Ali Khan", 100000)
savings.show_details()
savings.deposit(20000)
savings.withdraw(15000)
savings.calculate_profit()

print()

current = CurrentAccount("CA-2001", "Sara Ahmed", 50000)
current.show_details()
current.deposit(10000)
current.withdraw(5000)
current.calculate_profit()

print()

business = BusinessAccount("BA-3001", "Bilal Traders", 500000)
business.show_details()
business.deposit(100000)
business.withdraw(50000)
business.calculate_profit()
```
